Preparação do dataset para o Power BI
Fizemos isso por 3 motivos:
1. Reduzir o tamanho — o arquivo original tem 27 colunas mas para o dashboard só precisamos de 16. Menos colunas = arquivo menor = Power BI carrega mais rápido.
2. Corrigir o decimal — o arquivo original usa vírgula como decimal (ex: 630,5), o que confunde o Power BI e faz ele ler 630,5 como 6305. Salvamos com ponto decimal (ex: 630.5) para o Power BI interpretar os números corretamente.
3. Aproveitar as colunas novas — meus colegas criaram Media_Objetivas, Media_Final e Nivel durante a limpeza, então incluí essas colunas que vão deixar o dashboard mais rico e completo.

```python
df = pd.read_csv("Dados/dataset_limpo.csv")

colunas = ['Sigla_Estado', 'Municipio', 'Nota_CN', 'Nota_CH',
           'Nota_LC', 'Nota_MT', 'Nota_Redacao', 'Lingua_Estrangeira',
           'Status_Redacao', 'Presenca_CN', 'Presenca_CH',
           'Presenca_LC', 'Presenca_MT', 'Media_Objetivas', 
           'Media_Final', 'Nivel','Regiao','Nivel']

caminho = 'Dados/dataset_powerbiDadosLimpos.csv'

if not os.path.exists(caminho):
    df[colunas].to_csv(caminho, index=False, encoding='utf-8-sig')
    print("Dataset exportado! Arquivo: Dados/dataset_powerbiDadosLimpos.csv")
    display(df[colunas].head())
else:
    print("Arquivo já existe! Exportação ignorada.")

print(f"Arquivo salvo! {df.shape[0]:,} linhas, {len(colunas)} colunas")
```
