```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("Dados/dataset_limpo.csv")
pd.set_option('display.max_columns', None)
df_machine = df.copy()
```

```python
df_machine
```

```python
y = df_machine["Regiao"]  # Target

X = df_machine[["Nivel", "Lingua_Estrangeira"]] #preditora
```

OneHotEncoder


```python
X = pd.get_dummies(X, columns=["Nivel","Lingua_Estrangeira"], drop_first=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)
```

DECISION TREE

```python
print("=== ÁRVORE DE DECISÃO ===")
dt = DecisionTreeClassifier(random_state=42, max_depth=10, min_samples_split=50)
dt.fit(X_train, y_train_enc)
y_pred_dt = dt.predict(X_test)
print(f"Acurácia: {accuracy_score(y_test_enc, y_pred_dt):.4f}")
print(classification_report(y_test_enc, y_pred_dt, target_names=le.classes_))
```

RANDOM FOREST

```python
print("\n=== RANDOM FOREST ===")
rf = RandomForestClassifier(random_state=42, n_estimators=100, max_depth=10)
rf.fit(X_train, y_train_enc)
y_pred_rf = rf.predict(X_test)
print(f"Acurácia: {accuracy_score(y_test_enc, y_pred_rf):.4f}")
print(classification_report(y_test_enc, y_pred_rf, target_names=le.classes_))
```
